```python
import pandas as pd

df_vendas = pd.read_csv(r'C:\Users\Eiabe\Desktop\eletro\eletro_vendas.csv', encoding= 'latin-1', sep=';')
df_produtos = pd.read_csv(r'C:\Users\Eiabe\Desktop\eletro\eletro_produtos.csv')
df_clientes = pd.read_csv(r'C:\Users\Eiabe\Desktop\eletro\eletro_clientes.csv')
```

```python
df_vendas.isna().sum()
```

```python
df_vendas['quantidade'][pd.to_numeric(df_vendas['quantidade'], errors = 'coerce').isna()]
```

```python
duplicatas = df_vendas.columns.drop(['id_venda'])
df_vendas.duplicated(subset=duplicatas).sum()
```

```python
df_vendas['observacao'].str.contains(',').sum()
```

```python
df_vendas = df_vendas.drop_duplicates(subset=duplicatas, keep='first')
df_vendas = df_vendas.dropna(subset='quantidade')
df_vendas['quantidade'] = df_vendas['quantidade'].astype(int)

df_vendas['data_venda'] = pd.to_datetime(df_vendas['data_venda'])

df_vendas['observacao'] = df_vendas['observacao'].fillna("Sem Observação")

df_vendas = pd.merge(df_vendas, df_produtos, on = 'produto_id', how='left')
df_vendas = pd.merge(df_vendas, df_clientes, on='cliente_id', how='left')

display(df_vendas)
```
